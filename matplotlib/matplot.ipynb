```python
# Real data

# Oscar Winner Movies
oscar_movies = [
"The Dark Knight",
"The Hurt Locker",
"The King's Speech",
"The Artist",
"Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
plt.plot(years,oscar_revenue)
plt.title("Oscar movie Revenue in each year")
plt.xlabel("years")
plt.ylabel("Revenue")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar","Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 288, 829, 185, 275] # in $M
plt.plot(years, oscar_revenue, label = "Oscar")
plt.plot(years, non_oscar_movies,label="Non_oscar")
plt.title("Oscar and non Oscar  Movies Revenue ")
plt.xlabel("years")
plt.ylabel("Revenue")
plt.legend()

```

```python
award_movies = [
    "3 Idiots",
    "Dangal",
    "Baahubali 2",
    "RRR",
    "12th Fail"
]

years = [2009, 2016, 2017, 2022, 2023]

award_revenue = [460, 2024, 1810, 1250, 69]  # Revenue in ₹ Crore


popular_movies = [
    "Chennai Express",
    "PK",
    "Bajrangi Bhaijaan",
    "KGF Chapter 2",
    "Jawan"
]

popular_revenue = [423, 769, 969, 1200, 1160]  # Revenue in ₹ Crore

plt.plot(years, award_revenue,color="pink",marker = "o",linestyle="--",linewidth=4 ,label="Award Winning Movies")
plt.plot(years, popular_revenue, "<-g",label="Popular Movies")

plt.title("Indian Movies Revenue Comparison")
plt.xlabel("Year")
plt.ylabel("Revenue (₹ Crore)")
plt.legend()
plt.show()
```

```python
with plt.xkcd():
    plt.plot(years, award_revenue,label="Award Winning Movies")
    plt.plot(years, popular_revenue, label="Popular Movies")

    plt.title("Indian Movies Revenue Comparison")
    plt.xlabel("Year")
    plt.ylabel("Revenue (₹ Crore)")
    plt.style.use("default")
    plt.grid()
    plt.show()
```

```python
import numpy as np
oscar_movies = [
"The Dark Knight",
"The Hurt Locker",
"The King's Speech",
"The Artist",
"Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M

non_oscar_movies = ["Slumdog Millionaire", "Avatar","Inception", "Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M
years = [2008, 2009, 2010, 2011, 2012]
x =np.arange(0,len(years))
width = 0.4
plt.bar(x-width/2, oscar_revenue, width=width, label="Oscar Movies")
plt.bar(x+width/2, non_oscar_revenue, width= width, label="Non-Oscar Movies")
plt.xlabel("Years")
plt.title("Oscar and non Oscar  Movies Revenue ")
plt.ylabel("Revenue($M)")
plt.show()
```

```python
# Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
"Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
colors = ["red", "purple", "yellow", "blue", "pink", "red", "purple", "yellow", "blue", "pink"]
plt.scatter(age,bp,s=bp,cmap="OrRd")
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Prssure")
plt.colorbar()
plt.grid()
for i in range (len(people)):
    plt.annotate(people[i],xy=(age[i],bp[i]),xytext=(age[i],bp[i]))
```

```python
# Multiple Plots

cities = ["City A", "City B", "City C", "City D", "City E"]

# Winter: Temperature (C) vs Humidity (%)
winter_temp = [5, 2, 10, 0, 7]
winter_humidity = [80, 75, 65, 85, 70]

# Summer: Temperature (C) vs Humidity (%)
summer_temp = [25, 30, 28, 35, 27]
summer_humidity = [60, 50, 55, 45, 65]

plt.scatter(winter_temp, winter_humidity, label="Winter")
plt.scatter(summer_temp, summer_humidity, label="Summer")

plt.title("Summer vs Winter Data")
plt.xlabel("Temperature (℃)")
plt.ylabel("Humidity (%)")

plt.legend()
```

```python
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explode = [0, 0, 0, 0.2, 0]

plt.pie(
    amounts,
    labels=expenses,
    autopct="%1.1f%%",
    explode=explode,
    startangle=360,
    shadow=True,
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 1,
        "linestyle": "--"
    }
)

plt.title("Company Expenses Distribution")
plt.show()
```
