Using only Boolean Indexing:

1. Print passed students.

````python
students = np.array([78,45,91,23,67,88,39,59,95,72])
# Boss Challenge Solution
print("Here are the pass student name :-",students[students>40])
print("here are the list of failed Student:-",students[students<40])
print("Here is the studnet above 80:- ",students[students>80])
print("Below avrage marks:-",students[students<students.sum()/int(len(students))])
students[students<40]=40
print("Rplace the failed student with 40marks",students)
print("Numbr of distinction student",students[students>75])
print("No of student Below 50",students[students<50])
print("Even number only", students[students%2==0])
print("odd number only", students[students%2!=0])
print("Marks between 60 and 90",students[(students>60)&(students<90)])
````
