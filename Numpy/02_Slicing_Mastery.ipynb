## Quick Revision

### Slicing Syntax

```python
arr[start:stop:step]
```

- **start** → Included
- **stop** → Excluded
- **step** → Jump size
- Negative step means reverse direction.

# 🏆 Boss Challenge

```python
sales = np.array([1200,1400,1600,1800,2000,
                  2200,2400,2600,2800,3000])
```

Using **only slicing**, create:

1. First Quarter
2. Second Quarter
3. Weekend Sales
4. Reverse Sales
5. Alternate Sales
6. Middle Four Sales
7. Everything except first and last
8. Last Three Sales
9. Reverse first five sales
10. Reverse last five sales


````python
# Boss Challenge Solution
sales = np.array([1200,1400,1600,1800,2000,
                  2200,2400,2600,2800,3000])
print("First Quarter:- ",sales[0:int(len(sales)/4):])
print("Second Quarter:- ",sales[int(len(sales)/4):int(len(sales)/2)])
print("Weekly Sales:- ",sales[0:8:].sum())
print("Reverse Sales:- ",sales[::-1] )
print("Alternate Sales :- ",sales[0::2])
print("Middle four Sales:- ",sales[int(len(sales)/2-2):int(len(sales)/2+3)])
print("Except Last ans First: - ",sales[1:int(len(sales)-1):])
print("Reverse first 5 Sales :- ",sales[4::-1])
print("Reverse first 5 Sales :- ",sales[int(len(sales)):4:-1])
````
