# 🏆 Boss Challenge

Use only NumPy to:
1. Find total, average, min and max.
2. Identify above-average values.
3. Increase below-average values.
4. Compute cumulative sum.
5. Compute cumulative product.
6. Calculate standard deviation and variance.
7. Normalize the data.
8. Solve everything without loops.


```python
# Boss Challenge
marks = np.array([65,72,88,91,55,78,83,69,95,74])
sales = np.array([1200,1450,1320,1680,1750,1900,2100,2050,1980,2200])
print("Marks","\t","Sales")
print(marks.sum(),"\t",sales.sum())
print(marks.mean(),"\t",sales.mean())
sales[sales<sales.mean()] = sales.mean()
marks[marks<marks.mean()] = marks.mean()
print(marks,"\t",sales)
print(marks.cumsum(),"\t",sales.cumsum())
print(marks.cumprod(),"\t",sales.cumprod())
print(marks.std(),"\t",sales.std())
print(marks.var(),"\t",sales.var())
print(marks/marks.mean(),"\t",sales/sales.mean())
```
