# 4. Row Selection — `loc` and `iloc`

### Quick revision

- `iloc` → position-based selection
- `loc` → label/condition-based selection

Examples:

```python
data.iloc[0]
data.iloc[0:3]
data.loc[data["marks"] > 80]
```

### Practice

1. Select the first row using `iloc`.
2. Select the first 3 rows.
3. Select rows 2–4 and only `name` + `marks`.
4. Select all students whose marks are greater than 70.
5. Select students from the `CSE` branch.


````python
# Write your code here
print(data.iloc[0])
print(data.iloc[0:3])
print(data[["name","marks"]].iloc[2:5])
print(data.loc[data["marks"]>70])
print(data.loc[data["branch"]=="CSE"])

````

# 5. Boolean Filtering

### Quick revision

Comparison operators:

```python
>
<
>=
<=
==
!=
```

Combine conditions:

```python
(data["marks"] >= 70) & (data["age"] > 18)
```

Use `|` for OR.

### Practice

1. Students with marks >= 80.
2. Students with marks between 60 and 80.
3. CSE students with marks >= 70.
4. Students who are NOT from CSE.
5. Students older than 18 OR marks above 85.
6. Use `isin()` to select students from CSE or IT.


````python
# Write your code here
print(data.loc[data["marks"]>=80])
print(data.loc[(data["marks"]>60) & (data["marks"]<80)])
print(data.loc[(data["marks"]>=70) & (data["branch"]=="CSE")])
print(data.loc[data["branch"]!="CSE"])
print(data.loc[(data["age"]>18) & (data["marks"]>85)])
data[data["branch"].isin(["CSE", "IT"])]
````

### Practice

### Check yourself

A common mistake is confusing:

```python
data.drop_duplicates()
```

with:

```python
data.duplicated()
```

The first returns a cleaned DataFrame; the second identifies duplicates.


````python
students = {
    "name": ["Luffy", "Zoro", "Luffy", "Nami", "Zoro", "Sanji"],
    "branch": ["CSE", "IT", "CSE", "ECE", "IT", "CSE"],
    "age": [18, 19, 18, 20,  18, 20],
    "marks": [87, 92, 87, 78, 92, 85]
}

data = pd.DataFrame(students)

print(data.duplicated().count())
print(data.iloc[data.duplicated()])
drop = data.drop_duplicates()
print(drop)
print(f"{data.shape} before drop {drop.shape} after drop")

````

# 15. GroupBy — Core Data Analysis Skill

### Quick revision

`groupby()` is one of the most important Pandas operations for analytics.

Pattern:

```python

```

Other aggregations:

```python
.sum()
.mean()
.min()
.max()
.count()
```

### Practice

For a student dataset:

1. Find average marks by branch.
2. Find maximum marks by branch.
3. Find minimum marks by branch.
4. Count students by branch.
5. Find average age by branch.
6. Find the total marks by branch.


````python
print(data.groupby("branch")["marks"].mean())
print(data.groupby("branch")["marks"].max())
print(data.groupby("branch")["marks"].min())
print(data.groupby("branch")["age"].mean())
print(data.groupby("branch")["marks"].sum())
````

# 16. `groupby()` + `agg()` — Revision

### Quick revision

Multiple statistics can be calculated together:

```python
data.groupby("branch")["marks"].agg(["mean", "min", "max", "count"])
```

### Practice

Create a branch-wise summary containing:

- number of students
- average marks
- minimum marks
- maximum marks
- median marks

Then sort the result by average marks.


````python
print(data.groupby("branch")["name"].count().agg(["mean","min","max","count"]))
print(data.groupby("branch")["marks"].mean())
print(data.groupby("branch")["marks"].max())
print(data.groupby("branch")["marks"].min())
print(data.groupby("branch")["marks"].std())
````
